```python
files = ["book_01.txt", "book_02.txt", "book_03.txt", "book_04.txt", "book_05.txt"]


def get_words(file_name):
    text = open(file_name).read().lower()
    text = text.replace(".", "")
    return text.split()



def word_count_mapper(file_name):
    pairs = []
    for word in get_words(file_name):
        pairs.append((word, 1))
    return pairs


def word_count_reducer(pairs):
    counts = {}
    for word, one in pairs:
        if word in counts:
            counts[word] += one
        else:
            counts[word] = one
    return counts


all_pairs = []
for file_name in files:
    all_pairs += word_count_mapper(file_name)
word_counts = word_count_reducer(all_pairs)

print("WORD COUNT")
for word in sorted(word_counts):
    print(word, word_counts[word])



K = 10
top_words = sorted(word_counts.items(), key=lambda item: item[1], reverse=True)

print("\nTOP", K, "WORDS")
for word, count in top_words[:K]:
    print(word, count)


def index_mapper(file_name):
    pairs = []
    for word in get_words(file_name):
        pairs.append((word, file_name))
    return pairs


def index_reducer(pairs):
    index = {}
    for word, file_name in pairs:
        if word not in index:
            index[word] = []
        if file_name not in index[word]:      # keep the list unique
            index[word].append(file_name)
    return index


all_pairs = []
for file_name in files:
    all_pairs += index_mapper(file_name)
inverted_index = index_reducer(all_pairs)

print("\nINVERTED INDEX")
for word in sorted(inverted_index):
    print(word, inverted_index[word])
```
